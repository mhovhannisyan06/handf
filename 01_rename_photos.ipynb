```python
# 1) Connect Google Drive (a window will ask for permission: click Allow)
from google.colab import drive
drive.mount('/content/drive')

import os, csv, shutil

BASE = '/content/drive/MyDrive/handf_data'
RAW  = os.path.join(BASE, 'raw')
MAP  = os.path.join(BASE, 'rename_map.csv')
DRY_RUN = True   # True = only show the plan. Change to False to really do it.

files = sorted(f for f in os.listdir(BASE) if f.lower().endswith(('.jpg', '.jpeg', '.png')))
print(f'Found {len(files)} photos')

plan = [(old, f'page_{i:03d}.jpg') for i, old in enumerate(files, start=1)]
for old, new in plan:
    print(f'{old}  ->  raw/{new}')

if not DRY_RUN:
    os.makedirs(RAW, exist_ok=True)
    for old, new in plan:
        shutil.move(os.path.join(BASE, old), os.path.join(RAW, new))
    with open(MAP, 'w', newline='') as f:
        w = csv.writer(f)
        w.writerow(['original_name', 'file'])
        w.writerows(plan)
    print('Done!', len(plan), 'photos are now in raw/. Map saved to rename_map.csv')
```
